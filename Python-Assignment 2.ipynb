```python
#a.create age_list

age_list = [24,25,26,27,28]
```

```python
#b.create name_list

name_list = ["Rahul","Sam","Arjun","Meera","Anu"]
```

```python
#2.List Operations / modifications

#a.Append "Yazhini"
name_list.append("Yazhini")
print(name_list)
```

```python
#b.Insert 30 at index 2
age_list.insert(2,30)
print(age_list)
```

```python
#c.remove "Yazhini"
name_list.remove("Yazhini")
print(name_list)
```

```python
#d.pop last element from age_list
age_list.pop()
print(age_list)
```

```python
#e.Extend the age_list with additional ages [29, 30, 26].
age_list.extend([29,30,26])
print(age_list)
```

```python
#f.Sort age_list in descending order.
age_list.sort(reverse=True)
print(age_list)
```

```python
#g. Find Max age, Min age and sum of all ages from age_list.
max_age=max(age_list)
min_age=min(age_list)
total_age=sum(age_list)

print("Max age:",max_age)
print("Min age:",min_age)
print("Total age:",total_age)
```

```python
#3. Accessing List Elements:

#a. first element of name_list
print(name_list[0])
```

```python
#b.last element of name_list
print(name_list[-1])
```

```python
#c.elements from index 2 to index 4 in name_list
print(name_list[2:5])
```

```python
#d.elements in reverse order
print(name_list[::-1])
```

```python
#DICTIONARY

#a.dictionary creation
student_marks = {
    "Rahul": 85,
    "Sam": 76,
    "Arjun": 91,
    "Meera":68,
    "Anu":88
    }
print(student_marks)
```

```python
#b.Access and print the mark of a specific student, of your choice.
print(student_marks["Sam"])
```

```python
#c.Add "Janani" with amrk 80
student_marks["Janani"]=80
print(student_marks)
```

```python
#d. Update the mark of any one older student to 82.
student_marks["Meera"] = 82
print(student_marks)
```

```python
#e.Print all keys,values and key-value pairs

#print all keys
print(student_marks.keys())

#print all values
print(student_marks.values())

#print all key-value pairs
print(student_marks.items())
```

```python
#SETS

#a.set creation
my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)

#explanation:
#A set:
#does not keep duplicte elements
#contains unique values
#is unordred
```

```python
#b.change my_set
my_set[4] = 's'

#Explanation:a set is unordered and does not support indexing and duplication.
```

```python
#c.create 2 sets
set1 = {1,3,5,7,9}
set2 = {2,3,5,8,10}
```

```python
#d.Union and Intersection

union_set = set1 | set2
print(union_set)

intersection_set = set1 & set2
print(intersection_set)
```
