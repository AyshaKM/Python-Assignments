```python
#1. String Concatenation

string1 = "Hello"
name = input("Enter your name: ")
string2 = string1 + " " + name
print(string2)
```

```python
# Adding the third string

string3 = "welcome to Python programming"
string4 = string2 + ", " + string3
print(string4)
```

```python
#2. String Slicing and Indexing:

string4 = "Hello Zara, welcome to Python programming"

#a.first character
print("first character:",string4[0])
```

```python
#b.Last character
print("Last character:",string4[-1])
```

```python
#c. first 5 characters
print("first 5 characters:",string4[:5])
```

```python
#d.last 11 characters
print("last 11 characters:",string4[-11:])
```

```python
#e.string in reverse.
print("string in reverse:",string4[::-1])
```

```python
#f.print the word “Python”
print(string4[23:29])
```

```python
#3. String Methods:
strM = "Python beginner tutorial"

#a. to uppercase
print(strM.upper())
```

```python
#b.to lowercase
print(strM.lower())
```

```python
#c. Capitalize
print(strM.capitalize())
```

```python
#d. total number of occurrences of character ‘t’ in the string.
print(strM.count("t"))
```

```python
#e. Replace
print(strM.replace("Python","Machine Learning"))
```

```python
#tuples
t1 = (10,20,30)
t2 = (40,50,60)
```

```python
#a.concatenate
t_combine = t1 + t2
print(t_combine)
```

```python
#b.repeat 3 times
t_repeat = t_combine * 3
print(t_repeat)
```

```python
#c.access the 3rd element
print(t_combine[2])
```

```python
#d.access the first 3 elements
print(t_combine[0:3])
```

```python
#e.access the last 3 elements
print(t_combine[-3:])
```
